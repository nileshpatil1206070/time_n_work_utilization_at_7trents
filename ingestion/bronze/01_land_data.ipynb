```python

# 1. AUTOMATIC DATA LANDING - IMPORTS AND SETTINGS

# Purpose:
# 1. Define the GitHub-connected source folder.
# 2. Define the Unity Catalog landing location.
# 3. Prepare Python libraries for file operations.
# This cell does not copy or modify any files.


import os
import shutil

# Source: GitHub-connected Databricks workspace
github_base_path = ("/Workspace/Users/ankurkamboj.ynr@gmail.com/""time_n_work_utilization_at_7trents/"
    "data_ingest_github/")

# Destination: Unity Catalog landing volume
landing_base_path = ("/Volumes/severntrent/bronze/raw_data/""autoloader_landing/")

print("Source:", github_base_path)
print("Landing:", landing_base_path)
```

```python

# 2. AUTOMATIC DATA LANDING FUNCTION

# Purpose:
# 1. Discover dated folders in the GitHub-connected source.
# 2. Find CSV files inside each folder.
# 3. Copy only files not already present in the landing area.
# 4. Preserve the original date-wise folder structure.

# This function does not modify existing landed files.


def land_new_files():

    # 1. Find all dated folders
    date_folders = sorted(os.listdir(github_base_path))

    # 2. Process each dated folder
    for folder in date_folders:

        source_folder = os.path.join(github_base_path, folder)

        # Skip anything that is not a dated folder
        if not os.path.isdir(source_folder):
            continue

        # 3. Check CSV files inside the folder
        for file_name in os.listdir(source_folder):

            if not file_name.endswith(".csv"):
                continue

            source_file = os.path.join(source_folder, file_name)

            if not os.path.isfile(source_file):
                continue

            # 4. Define the matching landing folder
            destination_folder = os.path.join(landing_base_path, folder)

            destination_file = os.path.join(destination_folder, file_name)

            # 5. Skip files already landed
            if os.path.exists(destination_file):
                print("Already landed:", folder, "/", file_name)
                continue

            # 6. Create destination folder and copy new file
            os.makedirs(destination_folder, exist_ok=True)

            shutil.copyfile(source_file, destination_file)

            print("New file landed:", folder, "/", file_name)

    print("Data landing completed")
```

```python

# 3. EXECUTE AUTOMATIC DATA LANDING

# Purpose:
# 1. Discover available dated folders automatically.
# 2. Copy CSV files that haven't been landed yet.
# 3. Preserve the existing date-wise folder structure.
# 4. Skip files already present in the landing area.


land_new_files()
```
